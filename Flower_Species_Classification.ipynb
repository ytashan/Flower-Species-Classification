```python
# import libraries
import numpy as np
import pandas as pd
```

```python
# Dataset Load
from sklearn.datasets import load_iris
```

```python
#Show Data from the loaded dataset
load_iris()
```

```python
# Data Frame
flower = pd.DataFrame(load_iris().data)
flower
```

```python
#More Professional Column name changed by Flower Measure's
flower.columns = load_iris().feature_names
flower.head(5)
```

```python
# Sepearte Independent features and Dependent featutes
x = flower # x is all input data
y = load_iris().target # y is flower measurements (Target variables)
y
```

```python
# data Train (train test split)
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.30,random_state = 42)
#Means 70% data->training
#30% data testing
#random_state = 42 (The result is always same)
x_train
```

```python
x_test
```

```python
y_train
```

```python
y_test
```

```python
# Standardizing the data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
#x_train
x_test = scaler.transform(x_test)
```

```python
# inverse transform
scaler.inverse_transform(x_train)
```

```python
# Linear Regression Implementation
from sklearn.linear_model import LinearRegression
# Cross Validation
from sklearn.model_selection import cross_val_score
```

```python
# Models are training
regression = LinearRegression()
regression.fit(x_train,y_train)
```

```python
mse = cross_val_score(regression,x_train,y_train,scoring = 'neg_mean_squared_error',cv = 10)
#Data is divided into 10 parts
# Train-test is done 10 times
#Mean Squared error is output
#Note: neg_mean_squared_error means negative value will be output.

# Mean Calculate
print("Mean of MSE: ", np.mean(mse))
```

```python
# prediction
reg_pred = regression.predict(x_test)
reg_pred
```

```python
# Residual Distribution
import seaborn as sns
sns.displot(reg_pred-y_test,kind='kde')
```

```python
# R2 score
from sklearn.metrics import r2_score
score = r2_score(y_test,reg_pred)
print("R2 score: ",score)
```
